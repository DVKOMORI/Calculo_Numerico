# Cálculo Numérico — Atividade 5

Sistemas lineares e implementação numérica.

Um sistema linear com n equações e n incógnitas se escreve como A x = b. A matriz A guarda os coeficientes, o vetor x guarda as incógnitas e o vetor b guarda os termos independentes. A matriz aumentada [A | b] reúne A e b numa só tabela, e é nela que a eliminação de Gauss opera.

A eliminação zera os coeficientes abaixo de cada pivô com operações elementares nas linhas. Depois, a substituição regressiva lê a solução de baixo para cima. Sem pivoteamento, o algoritmo usa o elemento da diagonal como pivô e para se esse elemento for nulo.

### 2. Método da eliminação de Gauss

Sistema:

$$
\begin{cases}
2x + y - z = 8, \\
-3x - y + 2z = -11, \\
-2x + y + 2z = -3.
\end{cases}
$$

Matriz aumentada inicial:

$$
\left[\begin{array}{ccc|c}
2 & 1 & -1 & 8 \\
-3 & -1 & 2 & -11 \\
-2 & 1 & 2 & -3
\end{array}\right]
$$

**Etapa 1.** O pivô é a11 = 2.

- m21 = -3/2. A operação é L2 ← L2 + (3/2) L1.
- m31 = -1. A operação é L3 ← L3 + L1.

$$
\left[\begin{array}{ccc|c}
2 & 1 & -1 & 8 \\
0 & 1/2 & 1/2 & 1 \\
0 & 2 & 1 & 5
\end{array}\right]
$$

**Etapa 2.** O pivô é a22 = 1/2.

- m32 = 2 / (1/2) = 4. A operação é L3 ← L3 - 4 L2.

$$
\left[\begin{array}{ccc|c}
2 & 1 & -1 & 8 \\
0 & 1/2 & 1/2 & 1 \\
0 & 0 & -1 & 1
\end{array}\right]
$$

**Substituição regressiva.**

- Terceira equação: -z = 1, logo z = -1.
- Segunda equação: (1/2) y + (1/2)(-1) = 1, logo (1/2) y = 3/2 e y = 3.
- Primeira equação: 2x + 3 - (-1) = 8, logo 2x = 4 e x = 2.

A solução é (x, y, z) = (2, 3, -1).

Conferência nas equações originais:

- 2(2) + 3 - (-1) = 8
- -3(2) - 3 + 2(-1) = -11
- -2(2) + 3 + 2(-1) = -3

A célula abaixo repete a mesma eliminação com frações exatas.


```python
from fractions import Fraction


def _formatar_matriz(M):
    """Imprime a matriz aumentada, com a última coluna separada."""
    linhas = [[str(valor) for valor in linha] for linha in M]
    larguras = [
        max(len(linhas[i][j]) for i in range(len(linhas)))
        for j in range(len(linhas[0]))
    ]
    for linha in linhas:
        coeficientes = "  ".join(
            linha[j].rjust(larguras[j]) for j in range(len(linha) - 1)
        )
        termo = linha[-1].rjust(larguras[-1])
        print(f"[ {coeficientes} | {termo} ]")


def gauss_exato(A, b, mostrar=True):
    """Eliminação de Gauss sem pivoteamento, em aritmética racional.

    Mostra a matriz aumentada depois de cada pivô e devolve a solução.
    """
    n = len(b)
    M = [[Fraction(A[i][j]) for j in range(n)] + [Fraction(b[i])] for i in range(n)]

    if mostrar:
        print("Matriz aumentada inicial")
        _formatar_matriz(M)

    for k in range(n - 1):
        pivo = M[k][k]
        if pivo == 0:
            raise ZeroDivisionError(f"Pivô nulo na coluna {k + 1}.")
        for i in range(k + 1, n):
            fator = M[i][k] / pivo
            for j in range(k, n + 1):
                M[i][j] -= fator * M[k][j]
        if mostrar:
            print(f"\nDepois de eliminar abaixo do pivô da coluna {k + 1}")
            _formatar_matriz(M)

    x = [Fraction(0) for _ in range(n)]
    for i in range(n - 1, -1, -1):
        soma = sum(M[i][j] * x[j] for j in range(i + 1, n))
        x[i] = (M[i][n] - soma) / M[i][i]
    return x


A2 = [
    [2, 1, -1],
    [-3, -1, 2],
    [-2, 1, 2],
]
b2 = [8, -11, -3]

solucao = gauss_exato(A2, b2)
print("\nSolução exata:", [int(valor) for valor in solucao])
```

## Parte 2 — Propriedades e interpretação

### 3. Número de soluções

O critério usa os postos da matriz dos coeficientes e da matriz aumentada. Sejam n o número de incógnitas, r = posto(A) e r' = posto da matriz aumentada.

- Se r é menor que r', o sistema é inconsistente e não tem solução.
- Se r = r' = n, há solução única.
- Se r = r' e r é menor que n, há infinitas soluções, com n - r incógnitas livres.

**a)** Operação: L2 ← L2 - 2 L1.

$$
\left[\begin{array}{cc|c}
1 & 1 & 3 \\
2 & 2 & 6
\end{array}\right]
$$

passa a

$$
\left[\begin{array}{cc|c}
1 & 1 & 3 \\
0 & 0 & 0
\end{array}\right]
$$

A segunda linha virou 0 = 0. Resta uma equação independente, x + y = 3, para duas incógnitas. Então r = r' = 1, menor que 2: há **infinitas soluções**. Com y = t livre, x = 3 - t.

**b)** Operação: L2 ← L2 - 2 L1.

$$
\left[\begin{array}{cc|c}
1 & 1 & 3 \\
2 & 2 & 7
\end{array}\right]
$$

passa a

$$
\left[\begin{array}{cc|c}
1 & 1 & 3 \\
0 & 0 & 1
\end{array}\right]
$$

A segunda linha diz 0 = 1. O sistema é inconsistente: r = 1 e r' = 2. **Não há solução.**

**c)** Operações: L2 ← L2 - 2 L1 e depois L2 ← -(1/3) L2.

$$
\left[\begin{array}{cc|c}
1 & 1 & 3 \\
2 & -1 & 0
\end{array}\right]
$$

passa a

$$
\left[\begin{array}{cc|c}
1 & 1 & 3 \\
0 & -3 & -6
\end{array}\right]
$$

e depois a

$$
\left[\begin{array}{cc|c}
1 & 1 & 3 \\
0 & 1 & 2
\end{array}\right]
$$

Da segunda linha, y = 2. Da primeira, x + 2 = 3, logo x = 1. Os dois pivôs são não nulos, então r = r' = 2 = n: há **solução única**, (x, y) = (1, 2).


```python
import numpy as np


def classificar(A, b):
    """Classifica o sistema pelo posto de A e de [A|b]."""
    A = np.asarray(A, dtype=float)
    b = np.asarray(b, dtype=float).reshape(-1, 1)
    aumentada = np.hstack([A, b])
    posto_A = np.linalg.matrix_rank(A)
    posto_aug = np.linalg.matrix_rank(aumentada)
    n = A.shape[1]

    if posto_A < posto_aug:
        tipo = "nenhuma solução"
    elif posto_A == n:
        tipo = "solução única"
    else:
        tipo = f"infinitas soluções ({n - posto_A} incógnita livre)"
    return posto_A, posto_aug, tipo


sistemas = {
    "3a": ([[1, 1], [2, 2]], [3, 6]),
    "3b": ([[1, 1], [2, 2]], [3, 7]),
    "3c": ([[1, 1], [2, -1]], [3, 0]),
}

for nome, (A, b) in sistemas.items():
    posto_A, posto_aug, tipo = classificar(A, b)
    print(f"{nome}: posto(A) = {posto_A}, posto([A|b]) = {posto_aug} → {tipo}")
```

## Parte 3 — Implementação numérica em Python

### 4. Eliminação de Gauss sem pivoteamento

O algoritmo abaixo recebe A e b, monta a matriz aumentada [A | b], zera o que está abaixo de cada pivô e faz a substituição regressiva. Se algum pivô for nulo, ele interrompe: sem troca de linhas, o método não segue. O teste usa o sistema do exercício 2 e compara o vetor obtido com `numpy.linalg.solve`.


```python
def eliminacao_gauss(A, b, tol=1e-12):
    """Resolve Ax = b por eliminação de Gauss sem pivoteamento.

    Devolve a solução e a matriz aumentada já triangularizada.
    """
    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float).reshape(-1)
    n = b.shape[0]
    if A.shape != (n, n):
        raise ValueError("A deve ser quadrada e ter o mesmo tamanho de b.")

    # Matriz aumentada [A | b]. A cópia evita alterar os dados de entrada.
    M = np.column_stack([A, b])

    # Eliminação: para cada coluna k, zera as entradas abaixo do pivô.
    for k in range(n - 1):
        pivo = M[k, k]
        if abs(pivo) <= tol:
            raise ZeroDivisionError(
                f"Pivô nulo na posição ({k + 1}, {k + 1}). "
                "Sem pivoteamento o algoritmo não troca linhas."
            )
        for i in range(k + 1, n):
            fator = M[i, k] / pivo
            M[i, k:] -= fator * M[k, k:]
            M[i, k] = 0.0

    # Substituição regressiva, da última equação para a primeira.
    x = np.zeros(n)
    for i in range(n - 1, -1, -1):
        pivo = M[i, i]
        if abs(pivo) <= tol:
            raise ZeroDivisionError(
                f"Pivô nulo na linha {i + 1}. O sistema pode não ter solução única."
            )
        soma = np.dot(M[i, i + 1 : n], x[i + 1 : n])
        x[i] = (M[i, n] - soma) / pivo

    return x, M


def _formatar_numerica(M):
    """Imprime uma matriz numérica no mesmo formato da versão exata."""
    linhas = [[f"{valor:g}" for valor in linha] for linha in np.asarray(M, dtype=float)]
    larguras = [
        max(len(linhas[i][j]) for i in range(len(linhas)))
        for j in range(len(linhas[0]))
    ]
    for linha in linhas:
        coeficientes = "  ".join(
            linha[j].rjust(larguras[j]) for j in range(len(linha) - 1)
        )
        termo = linha[-1].rjust(larguras[-1])
        print(f"[ {coeficientes} | {termo} ]")


x_gauss, triangular = eliminacao_gauss(A2, b2)
x_numpy = np.linalg.solve(np.array(A2, dtype=float), np.array(b2, dtype=float))
residuo = np.array(A2, dtype=float) @ x_gauss - np.array(b2, dtype=float)

print("Matriz aumentada depois da eliminação")
_formatar_numerica(triangular)
print("Solução (Gauss sem pivoteamento):", x_gauss)
print("Solução (numpy.linalg.solve):     ", x_numpy)
print("Diferença máxima entre os dois:   ", np.max(np.abs(x_gauss - x_numpy)))
print("Resíduo Ax - b:                   ", residuo)
```

A solução numérica é (2, 3, -1), a mesma da resolução manual e a mesma de `numpy.linalg.solve`. O resíduo A x - b fica na ordem do erro de arredondamento, então o vetor calculado satisfaz o sistema original dentro da precisão de ponto flutuante.

A versão sem pivoteamento é direta, mas exige que todo pivô encontrado seja diferente de zero. Nos sistemas do exercício 3a e 3b isso não acontece: a redução produz uma linha de zeros e o algoritmo deve parar, porque não há solução única para devolver.


## Parte 4 — Aplicação

### 5. Modelo de balanço térmico

As temperaturas T1, T2 e T3 satisfazem

$$
\begin{bmatrix}
4 & -1 & 0 \\
-1 & 4 & -1 \\
0 & -1 & 3
\end{bmatrix}
\begin{bmatrix}
T_1 \\
T_2 \\
T_3
\end{bmatrix}
=
\begin{bmatrix}
15 \\
10 \\
10
\end{bmatrix}.
$$

**a) Resolução manual.**

Matriz aumentada:

$$
\left[\begin{array}{ccc|c}
4 & -1 & 0 & 15 \\
-1 & 4 & -1 & 10 \\
0 & -1 & 3 & 10
\end{array}\right]
$$

O multiplicador da segunda linha é m21 = -1/4. A operação L2 ← L2 + (1/4) L1 produz a linha (0, 15/4, -1, 55/4). A terceira linha já tem zero na primeira coluna.

$$
\left[\begin{array}{ccc|c}
4 & -1 & 0 & 15 \\
0 & 15/4 & -1 & 55/4 \\
0 & -1 & 3 & 10
\end{array}\right]
$$

O multiplicador seguinte é m32 = -4/15. A operação L3 ← L3 + (4/15) L2 zera a segunda coluna e deixa o pivô 41/15 com termo independente 41/3.

$$
\left[\begin{array}{ccc|c}
4 & -1 & 0 & 15 \\
0 & 15/4 & -1 & 55/4 \\
0 & 0 & 41/15 & 41/3
\end{array}\right]
$$

Substituição regressiva:

- (41/15) T3 = 41/3, logo T3 = 5.
- (15/4) T2 - 5 = 55/4, logo (15/4) T2 = 75/4 e T2 = 5.
- 4 T1 - 5 = 15, logo T1 = 5.

As três temperaturas valem 5. A matriz é estritamente diagonal dominante (4 > 1, 4 > 1 + 1 e 3 > 1), portanto invertível, e essa solução é a única.

**b) Implementação.** A célula seguinte resolve o mesmo sistema com a função do exercício 4.


```python
A5 = [
    [4, -1, 0],
    [-1, 4, -1],
    [0, -1, 3],
]
b5 = [15, 10, 10]

temperaturas_exatas = gauss_exato(A5, b5, mostrar=False)
temperaturas, _ = eliminacao_gauss(A5, b5)
temperaturas_numpy = np.linalg.solve(np.array(A5, dtype=float), np.array(b5, dtype=float))
residuo5 = np.array(A5, dtype=float) @ temperaturas - np.array(b5, dtype=float)

print("Temperaturas (fração):", [int(t) for t in temperaturas_exatas])
print("Temperaturas (Gauss):  ", temperaturas)
print("Temperaturas (NumPy):  ", temperaturas_numpy)
print("Resíduo Ax - b:       ", residuo5)
```

No modelo, T1 = T2 = T3 = 5. Os três nós da rede ficam na mesma temperatura. Substituindo esse valor nas equações, os acoplamentos entre vizinhos e as fontes 15, 10 e 10 se equilibram ao mesmo tempo. O resíduo numérico é nulo na precisão exibida, em acordo com a solução manual.
